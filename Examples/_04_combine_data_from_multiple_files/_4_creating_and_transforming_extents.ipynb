```python
import geokit as gk
from geokit.core.get_test_data import get_test_data
from pathlib import Path
```

# Extent
The Extent object represents geographic extents of an area and exposes useful
methods which depend on those extents. 

```python
# 1. Direct definition (minX, minY, maxX, maxY)
# EPSG:4326 is standard WGS84 Lat/Lon
ext = gk.Extent(6.0, 50.0, 6.75, 51.25, srs=gk.srs.EPSG4326)

# 2. Extract boundaries in different formats
print(f"BBox (minX, minY, maxX, maxY): {ext.xyXY}")
print(f"Alternative format (minX, maxX, minY, maxY): {ext.xXyY}")

# 3. Transform to a different projection (e.g., LAEA Europe)
# We can pad the area by 2% and fit it to a 500m grid resolution
transformed_ext = ext.castTo(gk.srs.EPSG3035).pad(2, percent=True).fit(500)

print(f"Transformed Box: {transformed_ext.box}")
```

## Create Extent from different data types

```python
data_folder = Path.cwd().parents[1].joinpath("geokit", "data")
```

#### Create extent from a raster

```python
# Create an Extent from an existing Raster-File
ext_from_raster = ext.fromRaster(get_test_data(file_name="gsa-ghi-like.tif", data_cache_folder=data_folder))
ext_from_raster
```

#### Create extent from a vector

```python
# Get Extent of Vector-File
from geokit.core.get_test_data import get_test_data

ext_from_vector = ext.fromVector(
    get_test_data(
        file_name="Natura2000_aachenClipped.shp",
        data_cache_folder=data_folder,
    )
)
ext_from_vector
```

#### Create extent from a raster

```python
# Get Extent of a geom
aachen_pt = gk.geom.point((6.083, 52.775), srs=gk.srs.EPSG4326)
aachen_buffered_area = aachen_pt.Buffer(1)
ext_from_geom = ext.fromGeom(aachen_buffered_area)
ext_from_geom
```
