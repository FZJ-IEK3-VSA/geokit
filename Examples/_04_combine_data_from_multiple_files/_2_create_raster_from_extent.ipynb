```python
import geokit as gk
import numpy as np
```

```python
# Define the target extent
ext = gk.Extent([6.0, 50.0, 7.0, 51.5], srs=gk.srs.EPSG4326)

# Create a dummy data matrix (7 rows, 6 columns)
raster_matrix = np.array(
    [
        [1, 2, 3, 1, 2, 3],
        [2, 2, 2, 2, 2, 2],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
    ]
)

# Automatically compute pixel sizes to fit the matrix into the extent
dx, dy = ext.computePixelSize(raster_matrix.shape[1], raster_matrix.shape[0])

# Generate the physical TIFF file
ext.createRaster(pixelWidth=dx, pixelHeight=dy, data=raster_matrix, output="./test_raster_with_ext.tif")
```
