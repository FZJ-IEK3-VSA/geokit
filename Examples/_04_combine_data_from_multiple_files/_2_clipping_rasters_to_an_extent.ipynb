```python
import geokit as gk
from geokit.core.get_test_data import get_test_data
from pathlib import Path
```

```python
# 1. Define the clipping area (The Extent)
# Coordinates: Lon_min, Lat_min, Lon_max, Lat_max
ext = gk.Extent(6.0, 50.0, 6.75, 51.25, srs=gk.srs.EPSG4326)

# 2. Define path to a source raster
# (Using Geokit's test data for this example)
data_folder = Path.cwd().parents[1].joinpath("geokit", "data")
source_raster = get_test_data(file_name="gsa-ghi-like.tif", data_cache_folder=data_folder)

# 3. Warp (Clip) the source to the extent
# resampleAlg determines how pixel values are interpolated
clipped_raster = ext.warp(source=source_raster, resampleAlg="bilinear", pixelWidth=0.01, pixelHeight=0.01)

# 4. Visualize the result
gk.drawRaster(clipped_raster, figsize=(6, 6))
```
