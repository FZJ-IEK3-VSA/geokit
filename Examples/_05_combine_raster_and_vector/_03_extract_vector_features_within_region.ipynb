```python
import geokit as gk
from geokit.core.get_test_data import get_test_data
import matplotlib.pyplot as plt
import pathlib
```

#### This example demonstrates how to find features (like wind turbines) within your region and create boolean "proximity masks."

```python
data_cache_folder = pathlib.Path().cwd().parents[1].joinpath("geokit", "data")
```

```python
aachen_rm = gk.RegionMask.load(
    get_test_data(
        file_name="aachenShapefile.shp",
        data_cache_folder=data_cache_folder,
    ),
    srs=3035,
    pixelRes=20,
)

aachen_points = aachen_rm.extractFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
        data_cache_folder=data_cache_folder,
    )
)
aachen_points.head()
```

```python
aachen_rm.drawGeoms(aachen_points, figsize=(5, 6), hideAxis=True, color="b")
```
