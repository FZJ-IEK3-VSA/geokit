# Extract Vector and Features within Region

```python
import geokit as gk
from geokit.core.get_test_data import get_test_data
import geokit.core.vector
import geokit.core.geom
import matplotlib.pyplot as plt
import pathlib

import matplotlib.pyplot as plt
```

#### This example demonstrates how to find features (like wind turbines) within your region and create boolean "proximity masks."

```python
data_cache_folder = pathlib.Path().cwd().parents[1].joinpath("geokit", "data")
```

```python
aachen_rm = gk.RegionMask.load(
    get_test_data(
        file_name="aachenShapefile.shp",
        data_cache_folder=data_cache_folder,
    ),
    srs=3035,
    pixelRes=20,
)

aachen_points = aachen_rm.extractFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
        data_cache_folder=data_cache_folder,
    )
)
# aachen_points.head()
```

```python
turbine_placements = gk.core.vector.extractFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
        data_cache_folder=data_cache_folder,
    ),
)
```

```python
fig, axs = plt.subplots(ncols=2, figsize=(5, 6))
ax_handle_1=aachen_rm.drawGeoms(aachen_points, figsize=(5, 6),  color="b", ax=axs[0])

ax_handle_2=aachen_rm.drawGeoms(turbine_placements, figsize=(5, 6), color="b", ax=axs[1])
```

```python
center_box = geokit.core.geom.box((4.035 * 10**4, 3.07 * 10**4, 4.07 * 10 **6, 3.10 * 10 ** 6), srs=3035)

# # now extract all "Bundeslaender" that overlap with this box AND clip their geometries to the box extent
# # the parameter 'scaleAttrs' would allow to scale numeric attributes of the states linearly with area share (but we have none in the sample data)
turbines_clipped = geokit.core.vector.extractAndClipFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
        data_cache_folder=data_cache_folder,
    ),
    geom=center_box,
    scaleAttrs=None,
)
```
