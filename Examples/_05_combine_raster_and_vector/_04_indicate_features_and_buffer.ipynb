# Indicate features and Buffer

```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import geokit.core.srs
import geokit.core.regionmask
from geokit.core.get_test_data import get_test_data
```

```python
# Repository data cache
data_cache_folder = pathlib.Path().cwd().parents[1].joinpath("geokit", "data")

# Define a region-centered CRS (Aachen)
aachen_centered_srs = geokit.core.srs.centeredLAEA(6.083, 50.775)

path_to_aachen_shape_file = get_test_data(file_name="aachenShapefile.shp", data_cache_folder=data_cache_folder)

# Load RegionMask from example shapefile
aachen_rm = geokit.core.regionmask.RegionMask.load(
    region=path_to_aachen_shape_file,
    pixelRes=20,
    srs=aachen_centered_srs,
)
```

```python
path_to_turbine_placements = get_test_data("turbinePlacements.shp", data_cache_folder=data_cache_folder)
is_near = aachen_rm.indicateFeatures(
    source=path_to_turbine_placements,
    buffer=500,
    multiProcess=False,
)
```

```python
aachen_rm.drawImage(is_near, figsize=(5, 6), hideAxis=True, cmap="Reds")
```
