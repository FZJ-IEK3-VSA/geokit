```python
import geokit as gk
from geokit.core.get_test_data import get_test_data
```

```python
# 1. Setup SRS and Load RegionMask
aachen_srs = gk.srs.centeredLAEA(6.083, 50.775)
aachen_rm = gk.RegionMask.load(
    get_test_data("aachenShapefile.shp"),
    pixelRes=20,
    srs=aachen_srs
)

# 2. Inspect Spatial Metadata
print(f"Extent: {aachen_rm.extent}")
print(f"Resolution: {aachen_rm.pixelWidth} x {aachen_rm.pixelHeight}")
```
