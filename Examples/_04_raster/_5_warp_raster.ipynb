# Warp Raster

This example demonstrates how to extract data from a raster.

```python
# extract value at a given location by interpolating
import pathlib
import geokit as gk
from geokit.core.get_test_data import get_test_data

data_cache_folder = pathlib.Path().cwd().parent.parent.joinpath("geokit", "data")
```

```python
path_to_sample_raster_data = get_test_data(file_name="clc-aachen_clipped.tif", data_cache_folder=data_cache_folder)
a = gk.raster.rasterInfo(path_to_sample_raster_data)
print(a)

a.srs.ExportToPrettyWkt()
```

```python
# Warp
path_to_sample_raster_data = get_test_data(file_name="clc-aachen_clipped.tif", data_cache_folder=data_cache_folder)

gk.drawRaster(
    source=path_to_sample_raster_data,
    figsize=(6, 6),
)
warped_raster = gk.raster.warp(
    path_to_sample_raster_data,
    srs=gk.srs.EPSG4326,
    resampleAlg="near",
)

axes_handle = gk.drawRaster(warped_raster, figsize=(6, 4))
```
