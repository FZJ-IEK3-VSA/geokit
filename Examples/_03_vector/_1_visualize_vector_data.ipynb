```python
import geokit as gk
import numpy as np
import pandas as pd
import pathlib
from geokit.core.get_test_data import get_test_data
```

```python
# data_cache_folder = pathlib.Path().cwd().parent.joinpath("geokit", "data")
data_cache_folder = pathlib.Path().cwd().parent.parent.joinpath("geokit", "data")
path_to_shape_file = get_test_data(
    file_name="gadm36_DEU_1.shp",
    data_cache_folder=data_cache_folder,
)
# # %%
```

```python
world = gk.vector.extractFeatures(path_to_shape_file, srs=4326)
ax = gk.drawGeoms(world, srs=4326, figsize=(6, 9))
```

```python
aachen_line = gk.geom.line(
    [
        (5.6086412456393614e-11, 1.0605099246605318e-09),
        (5.5966859847770105e-11, 11124.48031893997),
        (5.584717712977507e-11, 22249.12252108337),
    ],
    srs=4326,
)
df = pd.DataFrame(data={"asd": [1, 2, 3], "dd": [5, 6, 7], "geom": [aachen_line, aachen_line, aachen_line]})
gk.drawGeoms(df, colorBy="asd", srs=4326, ax=ax)
```
