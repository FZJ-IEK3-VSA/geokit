```python
import geokit as gk
```

```python
# Define two overlapping extents
ext1 = gk.Extent(6.00, 50.00, 6.75, 51.25, srs=gk.srs.EPSG4326)
ext2 = gk.Extent(6.25, 50.25, 7.00, 51.50, srs=gk.srs.EPSG4326)

# Combine them: ext3 will represent the bounding box that covers BOTH
ext3 = ext1 + ext2
```

```python
# This 'box' property (converts Extent to an OGR Geometry) and
# is useful for plotting or spatial intersections

# Visualize (if in a Jupyter environment)
AxH = gk.drawGeoms([ext3.box], figsize=(6, 6), fc="None", ec="#E58776", lw=5)
AxH2 = gk.drawGeoms([ext1.box, ext2.box], ax=AxH.ax, fc="#9CB0EC", ec="#132763", lw=1)
```
