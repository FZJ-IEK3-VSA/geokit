## Making subregions

Making subregions with tileize. Deconstruct a given geometry into a set of tiled geometries.
Example for federal state Bavaria.

```python
import pathlib
import geokit.core.vector
import geokit.core.geom
from geokit.core.get_test_data import get_test_data 
# Load test data
```

```python
data_cache_folder = pathlib.Path().cwd().parent.parent.joinpath("geokit", "data")

gadm36_deu_1_path = get_test_data(
    file_name="gadm36_DEU_1.shp",
    data_cache_folder=data_cache_folder,
)

federal_states_germany = geokit.core.vector.extractFeatures(gadm36_deu_1_path)
tiled_geoms = list(geokit.core.geom.tileize(federal_states_germany.geom.iloc[0], zoom=8))
print(len(tiled_geoms))

ax_handle_tilize=geokit.core.geom.drawGeoms(tiled_geoms, figsize=(6, 6))

```
