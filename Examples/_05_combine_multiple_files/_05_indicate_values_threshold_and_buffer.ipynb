# Indicate Values of a Raster

The 'indicateValue' method can be used to return a mask that indicates a specific value or range of values in the original raster. This can be used to identify areas with specific elevation ranges in height raster data. In this example, we will identify regions with a specific range of global horizontal irradiance (GHI). It will also be demonstrated how to create a buffer around this zone.

## Load Required Modules and Libraries

```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import geokit.core.srs
import geokit.core.regionmask
import geokit.core.raster
import geokit.core.util
from geokit.core.get_test_data import get_test_data
```

## Create a RegionMask Based on The Aachen Shape File 

```python
# Repository data cache
data_cache_folder = pathlib.Path().cwd().parents[1].joinpath("geokit", "data")

# Define a region-centered CRS (Aachen)
aachen_centered_srs = geokit.core.srs.centeredLAEA(6.083, 50.775)

# Load RegionMask from example shapefile
aachen_rm = geokit.core.regionmask.RegionMask.load(
    get_test_data(file_name="aachenShapefile.shp", data_cache_folder=data_cache_folder),
    pixelRes=20,
    srs=aachen_centered_srs,
)
```

## Indicate GHI Values

The indicateValue() function returns an array indicating regions with a GHI between 2.9 and 2.905.

```python
irradiance_raster_file = get_test_data("gsa-ghi-like.tif", data_cache_folder=data_cache_folder)
mask_thr = aachen_rm.indicateValues(
    source=irradiance_raster_file,
    value=(2.9, 2.905),
    multiProcess=False,
)
```

## Indicate GHI Values with Buffer

The Indicate Value function returns an array indicating regions with a GHI between 2.9 and 2.905, surrounded by a 100 m buffer area.

```python
mask_buf = aachen_rm.indicateValues(
    source=irradiance_raster_file,
    value=(2.9, 2.905),
    buffer=100,
    multiProcess=False,
)

np.unique(mask_buf)
```

# Plot the Indicated Values

```python
fig, axes = plt.subplots(ncols=2, figsize=(12, 6))
ax_handle_1 = geokit.core.util.drawImage(mask_thr, ax=axes[0])

ax_handle_1.ax.set_title(label="Areas with 2.9<=GHI<=2.905")

ax_handle_2 = geokit.core.util.drawImage(mask_buf, ax=axes[1])
ax_handle_2.ax.set_title(label="Areas with 2.9<=GHI<=2.905 (buffer=100m)")
```
