# Create Raster from Regionmask

Similar to the Extent object, the RegionMask object can be used to create a raster based on the extent of the region mask from a NumPy array. Therefore, a raster is created from a random NumPy array using the RegionMask created from a shapefile of the city of Aachen.

## Import Required Libraries and Modules

```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import geokit.core.srs
import geokit.core.regionmask
import geokit.core.raster
from geokit.core.get_test_data import get_test_data
```

## Create RegionMask Instance from Shapefile

```python
# Repository data cache
data_cache_folder = pathlib.Path().cwd().parents[1].joinpath("geokit", "data")

# Define a region-centered CRS (Aachen)
aachen_centered_srs = geokit.core.srs.centeredLAEA(6.083, 50.775)

path_to_shape_file = get_test_data(file_name="aachenShapefile.shp", data_cache_folder=data_cache_folder)
# Load RegionMask from example shapefile
aachen_region_mask = geokit.core.regionmask.RegionMask.load(
    region=path_to_shape_file,
    pixelRes=20,
    srs=aachen_centered_srs,
)
```

## Create Raster from RegionMask

For demonstration purposes, this example uses a random NumPy array to create raster data. In a real application, however, the user must provide the proper data.

```python
# example_data = np.random(aachen_rm.mask, dtype=np.uint8)
np.random.seed(0)
example_data = np.random.rand(*aachen_region_mask.mask.shape)
out_path = "./rm_raster.tif"
aachen_region_mask.createRaster(data=example_data, output=out_path)

print("Wrote:", out_path)
```

## Plot the RegionMask Mask and the Output Raster 

As can be seen, the new raster on the left fills the entire region mask object.

```python
fig, axs = plt.subplots(ncols=2, figsize=(12, 6))
geokit.core.raster.drawRaster(out_path, ax=axs[0])
aachen_region_mask.drawMask(ax=axs[1])
```
