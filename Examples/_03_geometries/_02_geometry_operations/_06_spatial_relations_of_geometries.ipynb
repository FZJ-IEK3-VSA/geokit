# Spatial Relations of Geometries

These examples demonstrate how spatial relations between geometries can be investigated. This means that one geometry can be determined either:

1. Contains
2. Crosses
3. Overlaps
4. Contains: 

Another geometry.

## Create Geometries to Showcase

```python
import geokit.core.srs
import geokit.core.geom

import matplotlib.pyplot as plt
```

```python
aachen_pt = geokit.core.geom.point((6.083, 52.775), srs=geokit.core.srs.EPSG4326)
# Get the buffer around this point
aachen_buffered_area = aachen_pt.Buffer(1)
# Get a boundary of the buffered area
boundary_polygon = aachen_buffered_area.Boundary()
# Make a line arround the buffer that does not touch it
aachen_centered_srs = geokit.core.srs.centeredLAEA(6.083, 50.775)

outside_line = geokit.core.geom.line(
    [
        (5, 51.5),
        (7.2, 51.5),
        (7.2, 54),
        (5, 54),
        (5, 51.5),
    ],
    srs=geokit.core.srs.EPSG4326,
)
```

## Visualize the Geometries


```python
ax_handle = geokit.core.geom.drawGeoms(
    aachen_pt,
    figsize=(6, 6),
    color="red",
)
ax_handle = geokit.core.geom.drawGeoms(
    aachen_buffered_area,
    figsize=(6, 6),
    ax=ax_handle,
)
ax_handle = geokit.core.geom.drawGeoms(
    boundary_polygon,
    figsize=(6, 6),
    ax=ax_handle,
    color="blue",
)

ax_handle = geokit.core.geom.drawGeoms(
    outside_line,
    figsize=(6, 6),
    ax=ax_handle,
    color="green",
)
```

## Investiagte the Spatial Relation

```python
# test for containment
print(aachen_buffered_area.Contains(aachen_pt))
print(aachen_buffered_area.Contains(outside_line))
print(boundary_polygon.Buffer(0.2).Contains(aachen_buffered_area))
print(boundary_polygon.Buffer(0.2).Crosses(aachen_buffered_area))
print(boundary_polygon.Buffer(0.2).Touches(aachen_buffered_area))
print(boundary_polygon.Buffer(0.2).Overlaps(aachen_buffered_area))
```
