# Create Extent Objects

There are multiple ways to create extent objects, which can be useful when dealing with many different types of objects and geodata sources.

```python
import geokit.core.srs
import geokit.core.extent
import geokit.core.raster
import geokit.core.geom
from geokit.core.get_test_data import get_test_data
from pathlib import Path
```

## Define an Extent from Coordinates

One way to define an extent is to provide the Extent object with coordinates and a spatial reference system.

```python
# 1. Direct definition (minX, minY, maxX, maxY)
# EPSG:4326 is standard WGS84 Lat/Lon
ext = geokit.core.extent.Extent(6.0, 50.0, 6.75, 51.25, srs=geokit.core.srs.EPSG4326)
```

## Create Extent from Files and in Memory Data

You can create extents from various data and file types, including .tif files, vector files (.shp), and in-memory rasters.


#### Create Extent from a Raster File (.tif)

```python
ext_from_raster = ext.fromRaster(get_test_data(file_name="gsa-ghi-like.tif"))
ext_from_raster
```

#### Create Extent from a Vector File .shp

```python
# Get Extent of Vector-File

ext_from_vector = ext.fromVector(get_test_data(file_name="Natura2000_aachenClipped.shp"))
ext_from_vector
```

#### Create extent from in Memory Raster

```python
# Get Extent of a geom
aachen_pt = geokit.core.geom.point((6.083, 52.775), srs=geokit.core.srs.EPSG4326)
aachen_buffered_area = aachen_pt.Buffer(1)
ext_from_geom = ext.fromGeom(aachen_buffered_area)
ext_from_geom
```

## Transform Extent

As shown below, you can retrieve the coordinates of the extent in various formats and transform them into other spatial reference systems.

```python
# 2. Extract boundaries in different formats
print(f"BBox (minX, minY, maxX, maxY): {ext.xyXY}")
print(f"Alternative format (minX, maxX, minY, maxY): {ext.xXyY}")

# 3. Transform to a different projection (e.g., LAEA Europe)
# We can pad the area by 2% and fit it to a 500m grid resolution
transformed_ext = ext.castTo(geokit.core.srs.EPSG3035).pad(2, percent=True).fit(500)

print(f"Transformed Box: {transformed_ext.box}")
```
