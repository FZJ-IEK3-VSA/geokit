# Clip Raster to an Extent

This example shows how to clip a raster to a given extent. This is useful if you want to extract a subset based on a user-defined extent or the extent of another object.

```python
import geokit.core.srs
import geokit.core.extent
import geokit.core.raster
from geokit.core.get_test_data import get_test_data
from pathlib import Path
```

```python
# 1. Define the clipping area (The Extent)
# Coordinates: Lon_min, Lat_min, Lon_max, Lat_max
ext = geokit.core.extent.Extent(6.0, 50.0, 6.75, 51.25, srs=geokit.core.srs.EPSG4326)

# 2. Define path to a source raster
# (Using Geokit's test data for this example)
source_raster = get_test_data(file_name="gsa-ghi-like.tif")

# 3. Warp (Clip) the source to the extent
# resampleAlg determines how pixel values are interpolated
clipped_raster = ext.warp(source=source_raster, resampleAlg="bilinear", pixelWidth=0.01, pixelHeight=0.01)

# 4. Visualize the result
geokit.core.raster.drawRaster(clipped_raster, figsize=(6, 6))
```
