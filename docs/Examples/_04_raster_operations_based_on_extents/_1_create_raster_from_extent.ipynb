# Create Raster from an Extent

GeoKit allows you to create a raster from an Extent object. This is convenient because it eliminates the need to determine the pixel size and number, and it provides a NumPy array.

```python
import numpy as np
import geokit.extent
import geokit.srs
```

```python
# Define the target extent
ext = geokit.extent.Extent([6.0, 50.0, 7.0, 51.5], srs=geokit.srs.EPSG4326)

# Create a dummy data matrix (7 rows, 6 columns)
raster_matrix = np.array(
    [
        [1, 2, 3, 1, 2, 3],
        [2, 2, 2, 2, 2, 2],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
        [3, 2, 1, 3, 2, 1],
    ]
)

# Automatically compute pixel sizes to fit the matrix into the extent
dx, dy = ext.computePixelSize(raster_matrix.shape[1], raster_matrix.shape[0])

# Generate the physical TIFF file
ext.createRaster(pixelWidth=dx, pixelHeight=dy, data=raster_matrix, output="./test_raster_with_ext.tif")
```
