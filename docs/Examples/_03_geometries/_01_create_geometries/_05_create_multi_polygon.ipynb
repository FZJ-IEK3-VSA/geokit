# Create Multipolygon

This example shows to create, visualize and store a multipolygon using GeoKit. Similar to a multiline object it allows to store multiple polygons as a single geometry object.

## Import required packages

```python
import geokit.geom
import geokit.vector
import geokit.srs
```

## Create Polygon

```python
# Create a Polygon geometry from a list of coordinate tuples.

# Define a projection for the Aachen area
aachen_centered_srs = geokit.srs.centeredLAEA(6.083, 50.775)

# Create 2 polygons from an ordered list of xy-coordinate tuples
aachen_polygon_1 = geokit.geom.polygon(
    [
        (0, 0),
        (40000, 0),
        (40000, 30000),
        (20000, 50000),
        (0, 30000),
        (0, 0),
    ],
    srs=aachen_centered_srs,
)

aachen_polygon_2 = geokit.geom.polygon(
    [
        (40000, 0),
        (80000, 0),
        (80000, 30000),
        (60000, 50000),
        (40000, 30000),
        (40000, 0),
    ],
    srs=aachen_centered_srs,
)

# combine polygons into a multipolygon
polys = [aachen_polygon_1, aachen_polygon_2]
multi_poly = geokit.geom.flatten(polys)
```

## Visualize Polygon

```python
ax_handle = geokit.geom.drawGeoms(
    geoms=multi_poly,
    figsize=(4, 4),
)
```

## Store Multipolygon as Vector File

```python
# Simple vector creation
geokit.vector.createVector(geoms=multi_poly, output="example_multi_polygon.shp")
```
