## Get Boundary
Get the boundary of an object by: obj.Boundary()

```python
import geokit.srs
import geokit.geom
import matplotlib.pyplot as plt
```

```python
# Make a line
aachen_centered_srs = geokit.srs.centeredLAEA(6.083, 50.775)

house_polygon = geokit.geom.polygon(
    [
        (0, 0),
        (40000, 0),
        (20000, 20000),
        (40000, 40000),
        (0, 40000),
        (0, 0),
    ],
    srs=4326,
)
print(type(house_polygon))
print(house_polygon.GetGeometryName())
```

```python
# Get a boundary
boundary_polygon = house_polygon.Boundary()
print(type(boundary_polygon))
print(boundary_polygon.GetGeometryName())
```

```python
fig, axs = plt.subplots(ncols=2, nrows=1, figsize=(12, 6))
ax_handle = geokit.geom.drawGeoms(house_polygon, figsize=(6, 6), color="red", ax=axs[0])
ax_handle = geokit.geom.drawGeoms(
    boundary_polygon,
    figsize=(6, 6),
    ax=axs[1],
    color="black",
)
```
