# Get Centroid

A centroid can be obtained from any geometry object. This example shows how to obtain the centroid of a line and a multipolygon object, but the process works for other geometries as well.

## Get Centroid from Line




```python
import geokit.srs
import geokit.geom
import osgeo.ogr
```

### Create a Line Geometry 

```python
# Create a LineString geometry from a list of coordinate tuples.

# Define a projection for Aachen area
aachen_centered_srs = geokit.srs.centeredLAEA(6.083, 50.775)

# Create a linestring geometry using a list of xy-coordinate tuples
house_line: osgeo.ogr.Geometry = geokit.geom.line(
    [
        (0, 0),
        (40000, 0),
        (40000, 30000),
        (20000, 50000),
        (0, 30000),
        (0, 0),
    ],
    srs=aachen_centered_srs,
)
```

### Apply Spatial Reference System to Centroid

It is important to apply the spatial reference system to the centroid. Otherwise, the coordinates cannot be interpreted correctly.

```python
house_centroid = house_line.Centroid()
house_centroid.AssignSpatialReference(aachen_centered_srs)
```

```python
# Plot the points
ax_handle = geokit.geom.drawGeoms(
    geoms=house_line,
    figsize=(4, 4),
    color="blue",
)
ax_handle = geokit.geom.drawGeoms(
    geoms=house_centroid,
    ax=ax_handle.ax,
    figsize=(4, 4),
    color="red",
)
```

## Get Centroid from Multipolygon Object

The same procedure is shown here for a multipolygon object.

### Create Multipolygon

```python
# Create a Polygon geometry from a list of coordinate tuples.

# Define a projection for the Aachen area
aachen_centered_srs = geokit.srs.centeredLAEA(6.083, 50.775)

# Create 2 polygons from an ordered list of xy-coordinate tuples
aachen_polygon_1 = geokit.geom.polygon(
    [
        (0, 0),
        (40000, 0),
        (40000, 30000),
        (20000, 50000),
        (0, 30000),
        (0, 0),
    ],
    srs=aachen_centered_srs,
)

aachen_polygon_2 = geokit.geom.polygon(
    [
        (40000, 0),
        (80000, 0),
        (80000, 30000),
        (60000, 50000),
        (40000, 30000),
        (40000, 0),
    ],
    srs=aachen_centered_srs,
)

# combine polygons into a multipolygon
polygons = [aachen_polygon_1, aachen_polygon_2]
multi_polygon = geokit.geom.flatten(polygons)
```

### Assign Spatial Reference System

```python
multi_polygon_centroid = multi_polygon.Centroid()
multi_polygon_centroid.AssignSpatialReference(aachen_centered_srs)
```

### Plot the Multipolygon and its Centroid

```python
# Plot the points
ax_handle = geokit.geom.drawGeoms(
    geoms=multi_polygon,
    figsize=(4, 4),
    color="blue",
)
ax_handle = geokit.geom.drawGeoms(
    geoms=multi_polygon_centroid,
    ax=ax_handle.ax,
    figsize=(4, 4),
    color="red",
)
```
