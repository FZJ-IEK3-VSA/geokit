# Extract Vector and Features within Region

This example demonstrates how to use a region mask to select vector data from a different shapefile and then clip it to the boundaries of the region mask. An imaginary box around the upper part of Aachen is therefore created. This box is then used to create a RegionMask instance, which extracts all wind turbine locations within that box.

## Load Libraries and Modules

```python
from geokit.core.get_test_data import get_test_data
import geokit.core.vector
import geokit.core.geom
import geokit.core.regionmask
import matplotlib.pyplot as plt
import pathlib

import matplotlib.pyplot as plt
```

## Create a Box Arround the Upper Part From Aachen

The boundary box is created iteratively by plotting the Aachen shapefile and examining the coordinates.

```python
# Get path to test data
path_to_aachen_shape_file = get_test_data(file_name="aachenShapefile.shp")

# boundary box tuple with (xmin, ymin, xmax, ymax)
srs_scale = 10**6
boundary_box_tuple = (4.035 * srs_scale, 3.07 * srs_scale, 4.07 * srs_scale, 3.105 * srs_scale)
box_upper_aachen = geokit.core.geom.box(boundary_box_tuple, srs=3035)

# Extract the Sha
aachen_shape_file = geokit.core.vector.extractFeatures(
    get_test_data(
        file_name="aachenShapefile.shp",
    ),
)

# Plot the aachen shape file and the box
ax_handle_aachen_shapefile = geokit.core.geom.drawGeoms(geoms=aachen_shape_file, srs=3035)
ax_handle_upper_box = geokit.core.geom.drawGeoms(geoms=box_upper_aachen, ax=ax_handle_aachen_shapefile, alpha=0.5)
```

## Extract All Turbines Within the Upper Aachen Boundary Box

```python
upper_aachen_region_mask = geokit.core.regionmask.RegionMask.load(
    box_upper_aachen,
    srs=3035,
    pixelRes=20,
)
turbine_locations_upper_aachen = upper_aachen_region_mask.extractFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
    )
)
type(turbine_locations_upper_aachen)
print(turbine_locations_upper_aachen)
```

## Extract all Turbines for Comparison

```python
turbine_placements_complete = geokit.core.vector.extractFeatures(
    get_test_data(
        file_name="turbinePlacements.shp",
    ),
)
```

## Plot Turbine Placements

The turbine placements extracted from the upper Aachen box region mask are now plotted alongside all turbine placements for comparison.

```python
fig, axs = plt.subplots(ncols=2, figsize=(14, 6))
ax_handle_1 = upper_aachen_region_mask.drawGeoms(turbine_locations_upper_aachen, ax=axs[0])
ax_handle_1.ax.set_title(label="Turbine placements within\nupper aachen region mask", fontsize=20)
# ax_handle_2 = upper_aachen_region_mask.drawGeoms(aachen_shape_file,  ax=axs[0])


ax_handle_2 = geokit.core.geom.drawGeoms(turbine_placements_complete, ax=axs[1], srs=3035)
ax_handle_2.ax.set_title(label="All turbine placements", fontsize=20)
# ax_handle_4 = geokit.core.geom.drawGeoms(aachen_shape_file, ax=axs[1], srs=3035)
```
