# Indicate features and Buffer

The indicateFeatures function operates in a similar way to the extractFeature function. The main difference is that it returns a raster of zeros and ones instead of a data frame containing geometry objects. A '1' indicates that the feature exists at that pixel, whereas a '0' indicates that no feature exists at that pixel.

## Import Modules and Libraries

```python
import pathlib
import numpy as np
import matplotlib.pyplot as plt
import geokit.srs
import geokit.regionmask
from geokit.get_test_data import get_test_data
```

## Create a Regionmask Object from a Shapefile

A RegionMask instance is created from the aachen shapefile.

```python
# Define a region-centered CRS (Aachen)
aachen_centered_srs = geokit.srs.centeredLAEA(6.083, 50.775)
path_to_aachen_shape_file = get_test_data(file_name="aachenShapefile.shp")

# Load RegionMask from example shapefile
aachen_region_mask = geokit.regionmask.RegionMask.load(
    region=path_to_aachen_shape_file,
    pixelRes=20,
    srs=aachen_centered_srs,
)
```

## Create a NumPy Array Indicating Turbine Locations

A NumPy array indicating the locations of the turbines is now created. A buffer zone of 500 metres is also set, in units of the SRS (metres in this case). Therefore, the area within a radius of 500 metres around the actual feature is also considered to be the feature.

```python
path_to_turbine_placements = get_test_data("turbinePlacements.shp")
turbine_location_raster = aachen_region_mask.indicateFeatures(
    source=path_to_turbine_placements,
    buffer=500,
    multiProcess=False,
)
```

## Plot the RegionMask

The turbine location raster and shapefile are drawn to visualise the result.

```python
ax_handle = aachen_region_mask.drawImage(turbine_location_raster, figsize=(5, 6), hideAxis=True, cmap="Reds")
```
